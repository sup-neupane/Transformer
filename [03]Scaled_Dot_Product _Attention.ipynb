# Scaled Dot Product Attention

The formula for Self-Attention is:

$$
\text{Attention}(Q, K, V) = \text{Softmax}(QK^T)V
$$

However, the actual formula used in Transformers is:

$$
\text{Attention}(Q, K, V) = \text{Softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V
$$

This is called **Scaled Dot Product Attention**.

---

## Why do we divide by \(\sqrt{d_k}\)?

Here:

- \(d_k\) = dimensionality of the key vector
- \(QK^T\) = dot product between query and key vectors

The scaling factor \(\sqrt{d_k}\) is used to prevent unstable gradients during training.

---

## Reason for Scaling

When the dimensionality is:

- **Low** → dot products have lower variance
- **High** → dot products can become very large, leading to high variance

Large dot product values create problems when passed through the Softmax function.

---

## Problem with Softmax

Very large input values make the Softmax output extremely peaked, causing it to produce probabilities very close to 0 or 1.

This leads to **vanishing gradients**, making training unstable and slowing learning.

---

## Solution

Dividing by:

$$
\sqrt{d_k}
$$

reduces the variance of the dot product values, keeping them in a stable range before applying Softmax.

This stabilizes training and improves gradient flow.
