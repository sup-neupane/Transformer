# Problem with Self-Attention

Consider the sentence:

> "A man saw the astronomer with a telescope."

This sentence is ambiguous because it can have two meanings:

1. A person saw the astronomer using a telescope.
2. A person saw an astronomer who was holding a telescope.

---

## Limitation of Self-Attention

A single Self-Attention mechanism generally focuses on one dominant relationship or interpretation at a time.

This means Self-Attention may capture only a single meaning or context of a sentence.

---

# Multi-Head Attention

Multi-Head Attention solves this problem by using multiple Self-Attention mechanisms in parallel.

Each Self-Attention module is called a **head**.

Different heads can learn different relationships and contextual meanings from the same sentence.

For example:

- One head may focus on grammatical relationships.
- Another head may focus on object associations.
- Another may focus on positional context.

---

## Multiple Weight Matrices

Each head has its own learned weight matrices:

- Query weight matrix: \(W_Q\)
- Key weight matrix: \(W_K\)
- Value weight matrix: \(W_V\)

Because each head uses different weights, each head learns different patterns and representations.

# Final Output of Multi-Head Attention

The outputs from all attention heads are concatenated together.

Then, the concatenated output is multiplied by another learned weight matrix \(W_O\).

This matrix combines information from all heads into a single representation.

![](./images/img14.png)

---

# Formula for Multi-Head Attention

$$
\text{MultiHead}(Q, K, V) =
\text{Concat}(\text{head}_1, \text{head}_2, \dots, \text{head}_h)W_O
$$

where:

$$
\text{head}_i =
\text{Attention}(QW_i^Q, KW_i^K, VW_i^V)
$$

Each head learns different contextual information from the same input sequence.
