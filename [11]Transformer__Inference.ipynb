# Transformer Decoder During Inference (Prediction) Time

The encoder behaves the same during both training and inference.

However, the decoder behaves differently:

- During training → non-autoregressive
- During inference → autoregressive

This is because during inference we do not have the output sentence available.

---

After the encoder finishes processing the input sentence, the decoder starts generating the output sequence.

---

# Step 1: Start Token

First, we send the `<START>` token to begin decoding.

The token is:

- embedded into a vector
- positional encoding is added

![](./images/img45.png)

---

# Step 2: Masked Multi-Head Attention

The embedding is passed into the Masked Multi-Head Attention block.

This converts the embedding into a contextual embedding.

The output is then:

- added with the original input embedding
- normalized using Layer Normalization

---

# Step 3: Cross Attention

The normalized output is sent to the Cross Attention block.

Cross Attention calculates attention scores using:

- encoder outputs
- decoder outputs from the previous layer

The output of cross attention is again:

- added with the output of masked multi-head attention
- normalized

![](./images/img46.png)

---

# Step 4: Feed Forward Neural Network

The output is passed into the Feed Forward Neural Network.

This introduces non-linearity into the model.

The output of the neural network is:

- added with the input of the network
- normalized again

---

# Step 5: Remaining Decoder Blocks

The same process is repeated for the remaining 5 decoder blocks.

---

# Step 6: Linear + Softmax Layer

The final output of the 6th decoder block is passed to a linear layer with:

$$
V
$$

neurons, where:

$$
V = \text{vocabulary size}
$$

Softmax converts these outputs into probabilities.

The word with the highest probability is selected.

---

# Step 7: Autoregressive Generation

The predicted word is again fed back into the decoder at the next timestamp.

This process repeats until the model predicts:

```text
<END>
```

# Note

Masking is also performed during inference.

Since masking is used during training, it must also be applied during prediction; otherwise, the model behavior during inference will differ from training, leading to poorer results.

Masking ensures that future tokens remain hidden, so the decoder only attends to previously generated tokens while predicting the next word.
