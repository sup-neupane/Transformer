# Cross Attention

Now we talk about the second Multi-Head Attention block in the decoder.

![](./images/img34.png)

In this block, the Query, Key, and Value vectors come from two different sources:

- From the decoder
- From the encoder

This mechanism is called **Cross Attention**.

---

# Definition

> "Cross-attention is a mechanism used in transformer architectures, particularly in tasks involving sequence-to-sequence tasks like translation or summarization. It allows the model to focus on different parts of the input sequence while generating the output sequence."

---

# Understanding with Translation Example

Consider the sentence:

```text
I like eating ice cream
```

The encoder processes the sentence and understands its contextual meaning.

The decoder then generates the translated sentence word by word at different timestamps.

---

# At a Particular Timestamp

Suppose the decoder is generating the word at timestamp 3.

The generated word depends on two things:

## 1. What has been generated till now

This information comes from the decoder’s **Masked Self-Attention** block.

It helps the decoder understand the previously generated words.

---

## 2. The original input sentence

This information comes from the encoder through **Cross Attention**.

Cross attention helps the decoder focus on the relevant parts of the input sentence while generating the current word.

---

# Key Idea

In Cross Attention:

- Query comes from the decoder
- Key and Value come from the encoder

This allows the decoder to “look at” the encoder output while generating text.

# Self-Attention vs Cross-Attention

![](./images/img35.png)

---

# Input

## Self-Attention
- Input is the embedding of a single sentence
- All tokens come from the same sequence

Example:
```text
Sentence → "I like ice cream"
```

---

## Cross-Attention
- Input comes from two different sequences:
  - Encoder output (original sentence)
  - Decoder output (generated sentence so far)

So it uses both:
- original input sentence
- partially generated output sentence

---

# Processing

## Self-Attention
- Query = Key = Value (same source)
- All come from the same sentence

So:
- each word attends to other words in the same sequence

---

## Cross-Attention
- Query comes from the **decoder (output side)**
- Key and Value come from the **encoder (input side)**

So:

- Query → what we are trying to generate now
- Key/Value → information from original input sentence

---

# Output

## Self-Attention Output

Gives:

> Contextual embedding of the same sentence

Meaning:
- each word understands its relationship with other words in the same sentence

---

## Cross-Attention Output (Important)

Cross-attention gives:

> Contextual embedding of the output sentence **conditioned on the input sentence**

---

# What does this mean?

Instead of just understanding a sentence, cross-attention tells:

> “What part of the input sentence is relevant for generating the next output word?”

---

### Example (Translation)

Input (encoder):
```text
I like ice cream
```

Output (decoder step-by-step):
```text
मलाई मन पर्छ ...
```

When generating each word, cross-attention helps the model:

- look back at the English sentence
- decide which word matters right now

---

# Intuition

| Type | Focus |
|------|------|
| Self-Attention | Relationships within the same sentence |
| Cross-Attention | Relationship between input sentence and output sentence |

---

# Simple Summary

- Self-attention = understand sentence internally  
- Cross-attention = align output with input meaning
