## Self-Attention Mechanism

One of the main challenges in NLP is transforming words into numerical representations (**vectorization**) so that neural networks can process them.

Some traditional methods are:

---

### One-Hot Encoding

| Word | mat | cat | rat |
|------|-----|-----|-----|
| mat  | 1   | 0   | 0   |
| cat  | 0   | 1   | 0   |
| rat  | 0   | 0   | 1   |

We can represent a sentence like:

`mat cat rat`

as:

`[100] [010] [001]`

However, one-hot encoding does not capture relationships or meanings between words.

---

### Bag of Words (BoW)

Example sentence:

`mat cat mat`

| Word Frequency | mat | rat | cat |
|----------------|-----|-----|-----|
| Count          | 2   | 0   | 1   |

In Bag of Words, words are represented based on their frequency in the sentence or document.

This method ignores word order and semantic meaning.

---

### Word Embedding

Modern NLP models use **word embeddings**.

Word embeddings convert words into dense numerical vectors using neural networks.

Unlike one-hot encoding, embeddings can capture **semantic meaning** and relationships between words.

Words with similar meanings tend to have similar vector representations.

## Problem with Word Embeddings

Consider the following sentences:

1. An apple a day keeps the doctor away  
2. Apple is healthy  
3. Apple is better than orange  
4. Apple makes great phones  

Suppose we represent each word using a **2-dimensional vector**:
- First dimension → Taste/Fruit-related meaning  
- Second dimension → Technology-related meaning  

A major problem with traditional word embeddings is that they generate a **single fixed vector** for a word by averaging its meaning across all contexts.

For example, if a dataset contains:
- 900 occurrences of **Apple** as a fruit  
- 100 occurrences of **Apple** as a company  

then the resulting embedding vector may look like:

`[0.9, 0.3]`

This means the representation is biased toward the fruit-related meaning.

Therefore, word embeddings depend heavily on the training data distribution and cannot properly capture different meanings of the same word in different contexts.


## Self-Attention

Self-attention solves the problem of static embeddings.

It is a mechanism that converts **static embeddings** into **contextual embeddings** by considering the relationship of a word with other words in the sentence.

This allows the same word to have different vector representations depending on its context.

For example, the word **"Apple"** in a technology-related sentence will have a different representation than in a fruit-related sentence.
