# Transformer Decoder

# Transformer Decoder During Training Time

There are **6 decoder blocks**, and all of them are identical in structure.

Each decoder block contains **3 main sub-blocks**:

1. Masked Multi-Head Attention  
2. Cross Attention  
3. Feed Forward Neural Network

# Transformer Decoder During Training

Suppose we have a machine translation task where we translate English to another language.

First, we discuss the training process only.

The English sentence first goes to the encoder and gets processed.

After that, the translated sentence (target language sentence) is given as input to the decoder.

---

# Input Block

The input block performs the following operations:

## 1. Shifting

The target sentence is shifted one position to the right, and a `<START>` token is added at the beginning.

---

## 2. Tokenization

The sentence is broken into tokens.

---

## 3. Embedding

Each token is converted into a vector of dimension **512**.

---

## 4. Positional Encoding

A positional encoding vector of dimension **512** is created and added to the embeddings.

---

Finally, we get the input vectors which are passed to the first decoder block.

![](./images/img38.png)

![](./images/img39.png)

---

# Decoder Block

Inside the decoder block, the following operations happen:

---

# 1. Masked Multi-Head Attention Block

For every input vector, contextual embeddings are created.

Future tokens are masked so that a token cannot attend to future words.

![](./images/img40.png)

---

# 2. Add and Normalize

The contextual embeddings are added to the original input through a residual connection.

The result is then layer normalized.

![](./images/img41.png)

---

# 3. Cross Attention Block (Multi-Head Attention)

Here, similarity scores are calculated between the input sequence and output sequence.

- Query vectors come from the decoder
- Key and Value vectors come from the encoder

This behaves similarly to self-attention.

The contextual embeddings from this block are added with the output of masked self-attention and then normalized.

![](./images/img42.png)

![](./images/img43.png)

---

# 4. Feed Forward Neural Network

This is similar to the feed forward network in the encoder.

It consists of:

- Two linear layers
- ReLU activation
- 2048 hidden neurons
- 512 output dimension

The vectors are processed as a batch (matrix).

This block introduces non-linearity into the model.

The output of the feed forward network is added to its input through a residual connection and normalized again.

This produces the final output of Decoder Block 1.

---

In the remaining decoder blocks, the same operations are repeated but with different learned parameters.

![](./images/img44.png)

---

# Output Block

The output block consists of two layers:

1. Linear Layer  
2. Softmax Layer  

---

# Linear Layer

The linear layer has:

$$
V
$$

neurons, where:

$$
V = \text{vocabulary size of the output language}
$$

It converts decoder outputs into vocabulary scores.

---

# Softmax Layer

Softmax converts these scores into probabilities for predicting the next word.
