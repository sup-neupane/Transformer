# Transformer Architecture (Encoder–Decoder)

The architecture consists of **encoder** and **decoder** blocks. The **Nx** notation represents that there are multiple identical layers stacked together.

![](./images/img24.png)

In *“Attention Is All You Need”*, there are **6 encoder and 6 decoder layers**. All these blocks are identical in structure.

---

# Encoder Architecture

The encoder has two main sub-blocks:

1. Self-Attention block  
2. Feed Forward Neural Network (FFN)

![](./images/img25.png)  
![](./images/img26.png)

# Example Sentence

Consider an example sentence:

**How are you**

---

This enters the input block that performs three operations:

## Tokenization  
Breaks the sentence into smaller units (tokens), usually words or subwords.

## Embedding  
Converts each word into a vector (numerical representation).  
The dimension of each vector is **512**.

## Positional Encoding  
A positional encoding vector is added to the embedding.  
The final vector size remains **512** after addition.

---

This is then passed into the first encoder block.

---

# Encoder Block Flow

The input is passed into the **Multi-Head Attention** block, which converts the input into contextual meaning.

Multi-head attention helps resolve ambiguous meanings in sentences.

The output of this block is a set of **contextual vectors**.

---

However, the input also bypasses the attention block through a residual connection.

So:

- The original input  
- The contextual output  

are added together.

![](./images/img27.png)

---

This combined output is then passed through **Layer Normalization**.

![](./images/img28.png)

# Feed Forward Neural Network Block

It is a two-layer neural network which accepts **512-dimensional inputs**, has **2048 hidden neurons** with ReLU activation, and outputs **512-dimensional vectors**.

![](./images/img29)

---

## Forward Pass

### First Layer

$$
Z_{norm} W_1 + b_1
$$

---

### Second Layer

$$
\text{ReLU}(Z_{norm} W_1 + b_1) W_2 + b_2
$$

---

The output of this feed forward neural network is again a set of vectors, each of **512 dimensions**.

![](./images/img30)

---

## Residual Connection and Normalization

The outputs \(Y_1, Y_2, Y_3\) are added with the bypassed outputs from the attention block:

- \(Z_{1norm}, Z_{2norm}, Z_{3norm}\)

This result is then normalized again to produce the final output of the first encoder block.

![](./images/img31)

---

These outputs are then passed as input to the second encoder block. The same process repeats again. The encoder block are same in terms of function/architecture but have different parameters. The final output of 6th encoder block is also of dimention 512.

# Why Use Residual Connections?

Residual connections improve performance, although the exact reason is not fully understood.

One intuition is that they help preserve the original input information by allowing the model to directly pass earlier representations forward.

---

# Why Use a Feed Forward Neural Network?

The exact reason is not fully clear, but the main purpose is to introduce **non-linearity** into the model.

This is important because multi-head attention and other components are mostly linear operations, and non-linearity increases the model’s expressive power.

---

# Why Use 6 Encoder Blocks?

A single encoder is not sufficient to fully understand complex human language.

Stacking multiple encoder layers allows the model to:

- build progressively richer representations
- capture deeper linguistic patterns
- improve understanding of context

This is why multiple encoder blocks are used instead of just one.
