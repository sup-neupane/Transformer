## Self-Attention

Self-attention is a mechanism that converts **static embeddings** into **contextual embeddings**.

It allows a word representation to change depending on the surrounding words in the sentence.

This helps the model understand context and relationships between words more effectively.

## How do we prepare Query, Key, and Value vectors?

This is done by **linear transformation** (multiplying embeddings by matrices).

Each embedding is multiplied with learned weight matrices to produce Query, Key, and Value vectors.

These matrices start as **random weights** initially. After making predictions, **backpropagation** adjusts these weights to improve performance.  
  
$$\text{Attention}(Q, K, V) = \text{Softmax}(QK^T)V$$
